# MODEL TRAINING 

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, roc_auc_score


# Features & target
X = df.drop('default_risk', axis=1)
y = df['default_risk']

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=77
)

X_train, X_test, y_train, y_test 
```

```python
# Scaling (for Logistic Regression)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

algorithm_names = []
accuracy_list = []
precision_list = []
recall_list = []
f1_list = []

def calculateClassificationMetrics(algorithm, predict, testY):

    acc = accuracy_score(testY, predict)
    prec = precision_score(testY, predict)
    rec = recall_score(testY, predict)
    f1 = f1_score(testY, predict)

    print(f"{algorithm} Accuracy: {acc:.2f}")
    print(f"{algorithm} Precision: {prec:.2f}")
    print(f"{algorithm} Recall: {rec:.2f}")
    print(f"{algorithm} F1 Score: {f1:.2f}")

    algorithm_names.append(algorithm)
    accuracy_list.append(acc)
    precision_list.append(prec)
    recall_list.append(rec)
    f1_list.append(f1)

    # Confusion Matrix Plot
    cm = confusion_matrix(testY, predict)
    plt.figure(figsize=(6,6))
    sns.heatmap(cm, annot=True, fmt='d')
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title(f"{algorithm} Confusion Matrix")
    plt.show()
```

```python
import os
os.makedirs("model", exist_ok=True)
```

```python
import os
import joblib
from sklearn.linear_model import LogisticRegression


model_path = 'model/logistic_model.pkl'

if os.path.exists(model_path):
    logistic_model = joblib.load(model_path)
    y_pred = logistic_model.predict(X_test)
    calculateClassificationMetrics("Logistic Regression", y_pred, y_test)
else:
    logistic_model = LogisticRegression(max_iter=1000)
    logistic_model.fit(X_train, y_train)
    y_pred = logistic_model.predict(X_test)
    
    joblib.dump(logistic_model, model_path)  
    
    print("Logistic Regression trained and model saved.")
    calculateClassificationMetrics("Logistic Regression", y_pred, y_test)
    
```

```python
import os
import joblib
from sklearn.ensemble import RandomForestClassifier

model_path = 'model/random_forest.pkl'

if os.path.exists(model_path):
    rf_model = joblib.load(model_path)
    y_pred = rf_model.predict(X_test)
    calculateClassificationMetrics("Random Forest", y_pred, y_test)
else:
    rf_model = RandomForestClassifier()
    rf_model.fit(X_train, y_train)
    y_pred = rf_model.predict(X_test)
    joblib.dump(rf_model, model_path)
    print("Random Forest trained and model saved.")
    calculateClassificationMetrics("Random Forest", y_pred, y_test)
```

```python
import os
import joblib
from xgboost import XGBClassifier

model_path = 'model/xgboost.pkl'

if os.path.exists(model_path):
    xgb_model = joblib.load(model_path)
    y_pred = xgb_model.predict(X_test)
    calculateClassificationMetrics("XGBoost", y_pred, y_test)
else:
    xgb_model = XGBClassifier()
    xgb_model.fit(X_train, y_train)
    y_pred = xgb_model.predict(X_test)
    joblib.dump(xgb_model, model_path)
    print("XGBoost trained and model saved.")
    calculateClassificationMetrics("XGBoost", y_pred, y_test)
```

# COMPARING THE MODEL'S RESULT

```python
import pandas as pd

columns = ["algorithm_name", "Accuracy", "Precision", "Recall", "F1 Score"]

values = []

for i in range(len(algorithm_names)):
    values.append([
        algorithm_names[i],
        round(accuracy_list[i], 2),
        round(precision_list[i], 2),
        round(recall_list[i], 2),
        round(f1_list[i], 2)
    ])

comparison_df = pd.DataFrame(values, columns=columns)

# Remove duplicates
comparison_df = comparison_df.drop_duplicates(subset="algorithm_name")

# Sort by best model
comparison_df = comparison_df.sort_values(by="F1 Score", ascending=False)

print("\nFinal Clean Classification Model Comparison Table:\n")
print(comparison_df.to_string(index=False))
```

```python
def plot_metric(metric_name):
    
    # Prepare values
    values = []
    for i in range(len(algorithm_names)):
        values.append([
            algorithm_names[i],
            accuracy_list[i],
            precision_list[i],
            recall_list[i],
            f1_list[i]
        ])
    
    # Column names
    columns = ["algorithm_name", "Accuracy", "Precision", "Recall", "F1 Score"]
    
    results_df = pd.DataFrame(values, columns=columns)

    # Remove duplicates (important for Jupyter 🔥)
    results_df = results_df.drop_duplicates(subset="algorithm_name")

    # Check metric exists
    if metric_name not in results_df.columns:
        print(f"Error: Metric '{metric_name}' not found.")
        return

    # Plot
    plt.figure(figsize=(8, 6))
    ax = sns.barplot(
        x="algorithm_name",
        y=metric_name,
        data=results_df,
        palette="viridis"
    )

    # Add values on bars
    for p in ax.patches:
        value = p.get_height()
        ax.annotate(
            f"{value:.3f}",
            (p.get_x() + p.get_width() / 2, value),
            ha='center',
            va='bottom',
            fontsize=10,
            color='black'
        )

    plt.title(f"{metric_name} Comparison Graph", fontsize=16)
    plt.xlabel("Algorithm", fontsize=13)
    plt.ylabel(metric_name, fontsize=13)
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()
```

```python
plot_metric("Accuracy")
```

```python
plot_metric("Precision")
```

```python
plot_metric("Recall")
```

```python
plot_metric("F1 Score")
```

```python
# Exclude target column
cols_to_scale = [col for col in df.columns if col != 'default_risk']

# Apply scaling
df[cols_to_scale] = scaler.fit_transform(df[cols_to_scale])
```

```python
# Load trained model
model_path = 'model/xgboost.pkl'   # or logistic_model.pkl
model = joblib.load(model_path)

# Predict
predictions = model.predict(df)

print("Predictions:")
print(predictions)
```

```python
# Probability of default (class 1)
probabilities = model.predict_proba(df)[:, 1]

print("Default Probabilities:")
print(probabilities)
```

```python
risk_levels = []

for prob in probabilities:
    if prob < 0.3:
        risk_levels.append("Low")
    elif prob < 0.7:
        risk_levels.append("Medium")
    else:
        risk_levels.append("High")

print("Risk Levels:")
print(risk_levels)
```

```python
df['prediction'] = predictions
df['default_probability'] = probabilities
df['risk_level'] = risk_levels

print(df.head())
```
